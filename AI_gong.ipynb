```python
from datetime import datetime
import random

# — 1. 경기 결과 입력 : 숫자는 반드시 int로 바꾼다 —
player     = input("선수 이름은? ")
place      = input("경기가 열린 곳은? ")
opponent   = input("상대 팀은? ")
goals      = int(input("골 수는? "))          # "2" -> 2
assists    = int(input("도움 수는? "))
score_us   = int(input("우리 팀 득점은? "))
score_them = int(input("상대 팀 득점은? "))

# — 2. 조사 고르기 : 받침이 있으면 '을/과', 없으면 '를/와' —
def josa(word, with_batchim, without_batchim):
    has_batchim = (ord(word[-1]) - 0xAC00) % 28 != 0
    return with_batchim if has_batchim else without_batchim

eul = josa(opponent, "을", "를")       # 리버풀을 / 맨시티를
gwa = josa(opponent, "과", "와")       # 리버풀과 / 맨시티와

# — 3. 문장 재료 : 같은 뜻의 문장을 여러 개 준비한다 —
OPENING = [
    "{player} 선수가 {place}에서 열린 {opponent}{gwa}의 경기에 선발 출전했습니다.",
    "{place}에서 펼쳐진 {opponent}{gwa}의 맞대결에 {player} 선수가 나섰습니다.",
]
RESULT = {
    "승": ["최종 스코어 {us} 대 {them}. {opponent}{eul} 꺾고 승점 3을 챙겼습니다."],
    "패": ["최종 스코어 {us} 대 {them}. {opponent}에게 아쉽게 무릎을 꿇었습니다."],
    "무": ["최종 스코어 {us} 대 {them}. 승부를 가리지 못했습니다."],
}

PLAY = {
    (True, True): ["{goals}골 {assists}도움으로 경기를 지배했습니다."],
    (True, False): ["{goals}골을 몰아치며 공격을 이끌었습니다."],
    (False, True): ["득점은 없었지만 {assists}개의 도움으로 힘을 보탰습니다."],
    (False, False): ["이날은 공격 포인트 없이 경기를 마쳤습니다."],
}

# — 4. 판단 : 점수를 비교해 승 / 패 / 무를 정한다 —
def judge(us, them):
    if us > them:
        return "승"
    elif us < them:
        return "패"
    else:
        return "무"

# — 5. 조립 : 재료를 하나씩 골라 기사 한 편을 만든다 —
def write_article():
    now = datetime.now().strftime("%Y-%m-%d %H:%M")   # 분 단위까지
    head = f"[프리미어 리그 속보 · {now}]"
    body = [
        random.choice(OPENING),
        random.choice(RESULT[judge(score_us, score_them)]),
        random.choice(PLAY[(goals > 0, assists > 0)]),
    ]
    article = " ".join(body).format(
        player=player, place=place, opponent=opponent,
        goals=goals, assists=assists,
        us=score_us, them=score_them, eul=eul, gwa=gwa)
    return f"{head}\n{article}"

news = write_article()
print(news)
```

```python
!pip -q install edge-tts

import edge_tts
from IPython.display import Audio, display

VOICE = "ko-KR-SunHiNeural"    # 여성 아나운서
# ko-KR-InJoonNeural         # 남성 아나운서
# ko-KR-HyunsuMultilingualNeural  # 남성 · 다국어

async def speak(text, filename="news.mp3", voice=VOICE, rate="+8%"):
    tts = edge_tts.Communicate(text, voice=voice, rate=rate)  # rate: 말하기 속도
    await tts.save(filename)                                  # mp3 파일로 저장
    return filename

mp3 = await speak(news)        # 코랩 셀에서는 await를 그대로 쓴다
display(Audio(mp3, autoplay=True))  # 브라우저에서 재생
```

```python
from datetime import datetime
import random

# 1. 경기 결과 입력 (전반/후반 분리)
player     = input("선수 이름은? ")
place      = input("경기가 열린 곳은? ")
opponent   = input("상대 팀은? ")
goals      = int(input("골 수는? "))
assists    = int(input("도움 수는? "))

us_first   = int(input("우리 팀 전반 득점은? "))
us_second  = int(input("우리 팀 후반 득점은? "))
them_first = int(input("상대 팀 전반 득점은? "))
them_second= int(input("상대 팀 후반 득점은? "))

score_us   = us_first + us_second
score_them = them_first + them_second

# 2. 조사 고르기
def josa(word, with_batchim, without_batchim):
    has_batchim = (ord(word[-1]) - 0xAC00) % 28 != 0
    return with_batchim if has_batchim else without_batchim

eul = josa(opponent, "을", "를")
gwa = josa(opponent, "과", "와")

# 3. 문장 재료 (2개씩 추가)
OPENING = [
    "{player} 선수가 {place}에서 열린 {opponent}{gwa}의 경기에 선발 출전했습니다.",
    "{place}에서 펼쳐진 {opponent}{gwa}의 맞대결에 {player} 선수가 나섰습니다.",
    "오늘 {place}에서는 {player} 선수가 {opponent}{gwa} 치열한 승부를 벌였습니다.",
    "{opponent}{gwa}의 경기가 열린 {place}에 {player} 선수가 모습을 드러냈습니다."
]

RESULT = {
    "승": [
        "최종 스코어 {us} 대 {them}. {opponent}{eul} 꺾고 승점 3을 챙겼습니다.",
        "경기는 {us} 대 {them}으로 마무리되며, 짜릿한 승리를 거두었습니다.",
        "상대를 완벽히 제압하며 {us} 대 {them}으로 승리의 기쁨을 맛봤습니다."
    ],
    "패": [
        "최종 스코어 {us} 대 {them}. {opponent}에게 아쉽게 무릎을 꿇었습니다.",
        "고군분투했으나 {us} 대 {them}으로 패하며 다음을 기약하게 되었습니다.",
        "상대의 거센 공세를 막지 못하고 {us} 대 {them}으로 아쉬운 패배를 기록했습니다."
    ],
    "무": [
        "최종 스코어 {us} 대 {them}. 승부를 가리지 못했습니다.",
        "치열한 공방전 끝에 {us} 대 {them}으로 무승부를 기록했습니다.",
        "양 팀 모두 총력을 다했으나 {us} 대 {them}으로 승점 1점씩 나눠 가졌습니다."
    ],
}

PLAY = {
    (True, True): [
        "{goals}골 {assists}도움으로 경기를 지배했습니다.",
        "공격 포인트를 휩쓸며 팀의 핵심 역할을 톡톡히 해냈습니다.",
        "맹활약을 펼치며 {goals}득점 {assists}도움이라는 훌륭한 기록을 남겼습니다."
    ],
    (True, False): [
        "{goals}골을 몰아치며 공격을 이끌었습니다.",
        "날카로운 결정력을 뽐내며 {goals}득점을 기록했습니다.",
        "팀의 득점을 책임지며 {goals}골을 성공시켰습니다."
    ],
    (False, True): [
        "득점은 없었지만 {assists}개의 도움으로 힘을 보탰습니다.",
        "이타적인 플레이로 {assists}도움을 기록하며 찬스를 만들었습니다.",
        "날카로운 패스로 {assists}개의 어시스트를 적립했습니다."
    ],
    (False, False): [
        "이날은 공격 포인트 없이 경기를 마쳤습니다.",
        "아쉽게도 득점이나 도움은 기록하지 못했습니다.",
        "활발히 그라운드를 누볐으나 공격 포인트를 올리진 못했습니다."
    ],
}

# 4. 판단 로직 (역전/추격 추가)
def judge(us, them):
    if us > them: return "승"
    elif us < them: return "패"
    else: return "무"

def match_flow(us_1, us_2, them_1, them_2):
    if us_1 < them_1 and (us_1 + us_2) > (them_1 + them_2):
        return "특히 전반전의 열세를 뒤집고 짜릿한 역전극을 만들어냈습니다."
    elif us_1 < them_1 and (us_1 + us_2) < (them_1 + them_2) and us_2 > them_2:
        return "후반 들어 맹렬히 추격했지만 경기를 뒤집기엔 역부족이었습니다."
    elif us_1 > them_1 and (us_1 + us_2) < (them_1 + them_2):
        return "전반전 리드를 지키지 못하고 뼈아픈 역전패를 당했습니다."
    return ""

# 5. 기사 조립
def write_article():
    now = datetime.now().strftime("%Y-%m-%d %H:%M")
    head = f"[프리미어 리그 속보 · {now}]"

    flow_sentence = match_flow(us_first, us_second, them_first, them_second)

    body = [
        random.choice(OPENING),
        random.choice(RESULT[judge(score_us, score_them)]),
        flow_sentence, # 흐름 문장 추가
        random.choice(PLAY[(goals > 0, assists > 0)])
    ]

    # 빈 문장 제거 후 조립
    body = [sentence for sentence in body if sentence]

    article = " ".join(body).format(
        player=player, place=place, opponent=opponent,
        goals=goals, assists=assists,
        us=score_us, them=score_them, eul=eul, gwa=gwa)
    return f"{head}\n{article}"

news = write_article()
print(news)
```
